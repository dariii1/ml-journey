```python
import torch
```

data definition

```python
torch.manual_seed(0)
x = torch.linspace(-1, 1, 100)
y = 3 * x + 2 + 0.1*torch.randn(100)
```

predictor coefficient

```python
# Defining a and b as zero tensors to evaluate the predict function 
a = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
# size step
lr = 0.1
```

predictor loop

```python
for step in range(100):
    y_pred = a * x + b                    # prediction function
    loss = ((y_pred - y) ** 2).mean()     # loss function
    loss.backward()                       # take a step back and calculate the difference eith respecct to the a and b

    with torch.no_grad():                 # using this function when we that so,e operation do not store anywhere
        a -= lr * a.grad                  #updating the data
        b -= lr * b.grad
    a.grad.zero_()
    b.grad.zero_()


    if step % 10 == 0:
        print(step, loss.item(), a.item(), b.item())

```

```python
print("--------------------------------------------------")
```

```python
def train(x, y, lr, steps=100):
    a = torch.tensor(0.0, requires_grad=True)
    b = torch.tensor(0.0, requires_grad=True)
    for _ in range(steps):
        loss = ((a * x + b - y) ** 2).mean()
        loss.backward()
        with torch.no_grad():
            a -= lr * a.grad
            b -= lr * b.grad
        a.grad.zero_()
        b.grad.zero_()
    return loss.item(), a.item(), b.item()

for lr in [0.01, 0.1, 1.5]:
    print(lr, train(x, y, lr))
```
