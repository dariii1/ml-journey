```python
import torch
from torch import nn
```

define the data

```python
torch.manual_seed(0)
x = torch.linspace(-1,1,100).unsqueeze(1) #reshape to (100, 1): 100 samples, 1 feature
y = 3 * x + 2 + 0.1 * torch.randn(100,1) 
```

define the model(using sgd model with different lr )

```python
model = nn.Linear(1, 1)
loss_fn = nn.MSELoss() #same work with loss definition from day 1
optimizer = torch.optim.SGD(model.parameters(), lr=0.1) #this part is same with update part of day one
```

```python
for i in range(100):
    y_pred = model(x)
    loss = loss_fn(y_pred, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if i % 20 == 0:
        print(i, loss.item())
```

parabola data

```python
torch.manual_seed(0)
y2 = 3 * x**2 + 2 + torch.randn(100, 1)
model = nn.Sequential(
nn.Linear(1, 16),
nn.Tanh(),
nn.Linear(16, 1),
)
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters())

for step in range(2000):
    y_pred = model(x)
    loss = loss_fn(y_pred, y2)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if step % 500 == 0:
        print(step, loss.item())

print("final loss:", loss.item())
```
